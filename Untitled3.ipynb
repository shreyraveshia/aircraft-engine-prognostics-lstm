```python
# Column names based on official C-MAPSS documentation
col_names = ['unit_number', 'time_cycles', 'setting_1', 'setting_2', 'setting_3'] + \
            [f'sensor_{i}' for i in range(1, 22)]

train_df = pd.read_csv('cmapss_data/CMaps/train_FD001.txt', sep=r'\s+', header=None, names=col_names)
test_df = pd.read_csv('cmapss_data/CMaps/test_FD001.txt', sep=r'\s+', header=None, names=col_names)
rul_df = pd.read_csv('cmapss_data/CMaps/RUL_FD001.txt', sep=r'\s+', header=None, names=['RUL'])

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("RUL shape:", rul_df.shape)

train_df.head(10)
```

```python
# Pick engine unit 1 (or try others: 24, 50, 77 — engines have different lifespans)
engine_id = 1
engine_data = train_df[train_df['unit_number'] == engine_id]

print(f"Engine {engine_id} lived for {engine_data['time_cycles'].max()} cycles before failure")

# Plot a few sensors known to show clear degradation trends (sensor 2, 3, 4, 7, 11, 15 are commonly informative)
sensors_to_plot = ['sensor_2', 'sensor_3', 'sensor_4', 'sensor_7', 'sensor_11', 'sensor_15']

fig, axes = plt.subplots(3, 2, figsize=(14, 10))
axes = axes.flatten()

for i, sensor in enumerate(sensors_to_plot):
    axes[i].plot(engine_data['time_cycles'], engine_data[sensor])
    axes[i].set_title(f'{sensor} over engine lifetime')
    axes[i].set_xlabel('Cycle')
    axes[i].set_ylabel(sensor)

plt.tight_layout()
plt.show()
```

```python
# Step 1: find the max cycle (failure point) for each engine
max_cycles = train_df.groupby('unit_number')['time_cycles'].max().reset_index()
max_cycles.columns = ['unit_number', 'max_cycle']

# Step 2: merge this back into train_df so every row knows its engine's failure point
train_df = train_df.merge(max_cycles, on='unit_number', how='left')

# Step 3: compute RUL = failure cycle - current cycle
train_df['RUL'] = train_df['max_cycle'] - train_df['time_cycles']

# Step 4: drop the helper column, we don't need it anymore
train_df.drop('max_cycle', axis=1, inplace=True)

# Sanity check: look at engine 1's RUL — should count DOWN to 0 at its last row
train_df[train_df['unit_number'] == 1][['unit_number', 'time_cycles', 'RUL']].tail(10)
```

```python
RUL_CAP = 125
train_df['RUL'] = train_df['RUL'].clip(upper=RUL_CAP)

# Sanity check: engine 1's RUL should now be flat at 125 early on, then count down normally later
train_df[train_df['unit_number'] == 1][['unit_number', 'time_cycles', 'RUL']].iloc[[0, 1, 2, 60, 61, 62, 180, 181, 182]]
```

```python
from sklearn.preprocessing import MinMaxScaler

# Columns we'll actually use as model input: settings + all sensors (NOT unit_number, time_cycles, or RUL — those aren't inputs)
feature_cols = ['setting_1', 'setting_2', 'setting_3'] + [f'sensor_{i}' for i in range(1, 22)]

scaler = MinMaxScaler()

# Fit ONLY on training data, then transform training data
train_df[feature_cols] = scaler.fit_transform(train_df[feature_cols])

# Transform test data using the SAME scaler (fit already happened above — do NOT re-fit here)
test_df[feature_cols] = scaler.transform(test_df[feature_cols])

# Sanity check: values should now roughly be between 0 and 1
train_df[feature_cols].describe().loc[['min', 'max']]
```

```python
# Find which feature columns had zero variance in training data (before scaling would've been better,
# but we can detect this by checking constant columns now)
constant_cols = [col for col in feature_cols if train_df[col].nunique() == 1]
print("Constant / zero-variance columns:", constant_cols)

# Drop them from our feature list (keep the dataframes intact, just track which columns we'll actually use)
feature_cols = [col for col in feature_cols if col not in constant_cols]
print(f"\nUsing {len(feature_cols)} features for modeling")
```

```python
SEQUENCE_LENGTH = 30

def create_sequences(df, feature_cols, sequence_length):
    sequences = []
    labels = []

    for unit in df['unit_number'].unique():
        engine_data = df[df['unit_number'] == unit].sort_values('time_cycles')
        feature_data = engine_data[feature_cols].values
        rul_data = engine_data['RUL'].values

        num_rows = len(engine_data)

        if num_rows < sequence_length:
            # Not enough history for this engine — skip it for now (rare edge case, we'll note it)
            continue

        for i in range(num_rows - sequence_length + 1):
            seq = feature_data[i : i + sequence_length]
            label = rul_data[i + sequence_length - 1]  # RUL at the END of the window
            sequences.append(seq)
            labels.append(label)

    return np.array(sequences), np.array(labels)

X_train, y_train = create_sequences(train_df, feature_cols, SEQUENCE_LENGTH)

print("X_train shape:", X_train.shape)   # should be (num_windows, 30, 17)
print("y_train shape:", y_train.shape)   # should be (num_windows,)
```

```python
# Check: does every test engine have at least 30 cycles?
test_lengths = test_df.groupby('unit_number')['time_cycles'].max()
print("Shortest test engine has", test_lengths.min(), "cycles")
print("Number of test engines with fewer than 30 cycles:", (test_lengths < SEQUENCE_LENGTH).sum())
```

```python
def create_test_sequences(df, feature_cols, sequence_length):
    sequences = []

    for unit in df['unit_number'].unique():
        engine_data = df[df['unit_number'] == unit].sort_values('time_cycles')
        feature_data = engine_data[feature_cols].values

        # take only the LAST `sequence_length` rows for this engine
        seq = feature_data[-sequence_length:]
        sequences.append(seq)

    return np.array(sequences)

X_test = create_test_sequences(test_df, feature_cols, SEQUENCE_LENGTH)
y_test = rul_df['RUL'].values  # the true answers, already loaded earlier

print("X_test shape:", X_test.shape)   # should be (100, 30, 17) — one window per test engine
print("y_test shape:", y_test.shape)   # should be (100,)
```
